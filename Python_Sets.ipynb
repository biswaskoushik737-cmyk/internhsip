# Python Sets

```python
# Creating a Python Set
my_set = {1, 2, 3, 4, 4, 5}  # Duplicates are automatically removed
print("Initial Set:", my_set)

# Adding elements
my_set.add(6)
print("After adding 6:", my_set)

# Removing elements
my_set.discard(3)
print("After removing 3:", my_set)

# Set Operations
set_a = {1, 2, 3}
set_b = {3, 4, 5}

print("\nSet Operations:")
print("Union:", set_a.union(set_b))          # Or set_a | set_b
print("Intersection:", set_a.intersection(set_b))  # Or set_a & set_b
print("Difference:", set_a.difference(set_b))      # Or set_a - set_b
```

```python
# Define a set
names_set = {"Alice", "Bob", "Charlie"}

# Check for membership using 'in'
is_alice_present = "Alice" in names_set
print(f"Is 'Alice' in the set? {is_alice_present}")

# Check for membership using 'not in'
is_david_absent = "David" not in names_set
print(f"Is 'David' NOT in the set? {is_david_absent}")
```
