# Pandas Data Summary

```python
import pandas as pd
import numpy as np

# Create a sample DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
    'Age': [25, 30, 35, np.nan, 28, 40],
    'Salary': [50000, 60000, np.nan, 80000, 55000, 95000],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'HR', 'IT'],
    'Join_Date': pd.date_range(start='2020-01-01', periods=6, freq='YE')
}
df = pd.DataFrame(data)

print("--- DataFrame Preview ---")
display(df.head())

print("\n--- Dataframe Info ---")
df.info()

print("\n--- Descriptive Statistics (Numerical) ---")
display(df.describe())

print("\n--- Descriptive Statistics (Categorical/Object) ---")
display(df.describe(include=['object']))

print("\n--- Missing Values Count ---")
print(df.isnull().sum())

print("\n--- Shape of DataFrame ---")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
```

```python
# Group the data by 'Department' and calculate the mean salary
mean_salary_by_dept = df.groupby('Department')['Salary'].mean().reset_index()

print("--- Mean Salary by Department ---")
display(mean_salary_by_dept)
```
