# Python Tuples

```python
# Python Tuples Introduction

# 1. Creating a Tuple
# Tuples are ordered, immutable, and allow duplicate values.
my_tuple = ("apple", "banana", "cherry", "apple")
print("Created Tuple:", my_tuple)

# 2. Accessing Elements
# You can access elements by indexing, starting from 0
print("First element:", my_tuple[0])
print("Last element:", my_tuple[-1])

# 3. Tuple Length
print("Tuple length:", len(my_tuple))

# 4. Tuples are Immutable
try:
    # This will raise a TypeError because tuples cannot be modified
    my_tuple[1] = "orange"
except TypeError as e:
    print("\nImmutability check:", e)

# 5. Tuple Methods (count and index)
print("\nOccurrences of 'apple':", my_tuple.count("apple"))
print("Index of 'cherry':", my_tuple.index("cherry"))
```
