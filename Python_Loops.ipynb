# Python Loops

```python
# 1. For Loop: Iterating over a list
fruits = ["apple", "banana", "cherry"]
print("Iterating over a list:")
for fruit in fruits:
    print(f" - {fruit}")

# 2. For Loop with range()
print("\nUsing range(3):")
for i in range(3):
    print(f" - Index: {i}")

# 3. While Loop
print("\nUsing a while loop:")
count = 0
while count < 3:
    print(f" - Count: {count}")
    count += 1

# 4. Loop Control: break and continue
print("\nUsing break and continue:")
for num in range(1, 6):
    if num == 2:
        continue  # Skip 2
    if num == 4:
        break     # Stop loop at 4
    print(f" - Number: {num}")
```
